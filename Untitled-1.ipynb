```python
#imports and folder set up
import os, random, math
from pathlib import Path
import numpy as np
from PIL import Image, ImageFilter, ImageOps
import matplotlib.pyplot as plt
from torchvision import datasets, transforms

# Create folders for both datasets
Path("real_images").mkdir(exist_ok=True)
Path("generated_images").mkdir(exist_ok=True) # Add your generated samples here
```

```python
# Download a small sample from CIFAR-10 (~200 images), preparing real image dataset
real_root = "real_images"
if len(os.listdir(real_root)) < 50:
    tfm = transforms.Compose([transforms.ToTensor()])
    cifar = datasets.CIFAR10(root="data", train=True, download=True, transform=tfm)
    idxs = np.random.choice(len(cifar), size=200, replace=False)

    for i, idx in enumerate(idxs):
        img, _ = cifar[idx]
        pil = transforms.ToPILImage()(img).resize((256,256), Image.BICUBIC)
        pil.save(os.path.join(real_root, f"real_{i:03d}.png"))

print("Real images:", len(os.listdir(real_root)))
```

```python
#add or simulate generated images for testing
gen_root = "generated_images"
if len(os.listdir(gen_root)) == 0:
    # Create synthetic color blocks as placeholder outputs
    colors = [(220,50,50),(50,180,90),(60,90,200),(200,160,40)]
    for i, color in enumerate(colors * 50): # ~200 fake samples
        img = Image.new("RGB", (256,256), color)
        img.save(os.path.join(gen_root, f"gen_placeholder_{i:03d}.png"))

print("Generated images:", len(os.listdir(gen_root)))
```

```python
# standardise image sizes for both folders
def resize_folder(in_dir, size=(256,256)):
    for fname in os.listdir(in_dir):
        path = os.path.join(in_dir, fname)
        try:
            img = Image.open(path).convert("RGB")
            if img.size != size:
                img = img.resize(size, Image.BICUBIC)
                img.save(path)
        except Exception as e:
            print("Skipped:", fname, "-", e)

resize_folder("real_images")
resize_folder("generated_images")
```

```python
#visual check
def show_grid(folder, n=8):
    files = [f for f in os.listdir(folder) if f.lower().endswith((".png",".jpg",".jpeg"))]
    if not files:
        print("No images in", folder); return
    sample = random.sample(files, min(n, len(files)))
    cols, rows = 4, math.ceil(len(sample)/4)
    plt.figure(figsize=(12, 3*rows))
    for i, f in enumerate(sample, 1):
        img = Image.open(os.path.join(folder, f))
        plt.subplot(rows, cols, i)
        plt.imshow(img); plt.axis("off")
    plt.suptitle(folder, fontsize=12)
    plt.tight_layout(); plt.show()

show_grid("real_images")
show_grid("generated_images")
```

```python
#final validation
real_count = len([f for f in os.listdir("real_images") if
    f.lower().endswith((".png",".jpg",".jpeg"))])
gen_count = len([f for f in os.listdir("generated_images") if
    f.lower().endswith((".png",".jpg",".jpeg"))])

print("Real images count:", real_count)
print("Generated images count:", gen_count)
assert real_count > 0 and gen_count > 0, "Please ensure both folders contain images."
```

```python
import torch, numpy as np
from PIL import Image
import open_clip

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

# Load pretrained CLIP model and preprocessing transforms
model, _, preprocess = open_clip.create_model_and_transforms(
    "ViT-B-32", pretrained="laion2b_s34b_b79k", device=device
)
tokenizer = open_clip.get_tokenizer("ViT-B-32")

def clipscore_openclip(references, predictions):
    """Compute cosine similarity between text and image pairs."""
    assert len(references) == len(predictions), "Text and image lists must match in length."
    scores = []
    with torch.no_grad():
        for txt, img_path in zip(references, predictions):
            # Encode text
            text_tok = tokenizer([txt]).to(device)
            text_emb = model.encode_text(text_tok)
            text_emb /= text_emb.norm(dim=-1, keepdim=True)

            # Encode image
            image = Image.open(img_path).convert("RGB")
            image_tensor = preprocess(image).unsqueeze(0).to(device)
            img_emb = model.encode_image(image_tensor)
            img_emb /= img_emb.norm(dim=-1, keepdim=True)

            # Compute cosine similarity (map from [-1,1] [0,1])
            cos_sim = (text_emb @ img_emb.T).item()
            scores.append((cos_sim + 1.0) / 2.0)

    return {"scores": scores, "average_score": float(np.mean(scores))}

# Example usage (replace with your own prompts and generated images)
references = ["a red vehicle", "a small animal"]
predictions = ["generated_images/gen_000.png", "generated_images/gen_001.png"]

clip_res = clipscore_openclip(references, predictions)
print("CLIP Similarity Results:", clip_res)
```
